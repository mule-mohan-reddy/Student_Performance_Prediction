```python
# 5. Data Preprocessing
X = df.drop("G3", axis=1)
y = df["G3"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# 5. Data Preprocessing
categorical_columns = X.select_dtypes(include="object").columns.tolist()
numerical_columns = X.select_dtypes(exclude="object").columns.tolist()

print("Categorical columns:")
print(categorical_columns)

print("\nNumerical columns:")
print(numerical_columns)
```

```python
# 5. Data Preprocessing
X_encoded = pd.get_dummies(X, columns=categorical_columns, drop_first=True)

print("Original shape:", X.shape)
print("Encoded shape:", X_encoded.shape)
```

```python
# 6. Train-Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# 7. Linear Regression
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
y_pred = model.predict(X_test)

print("Predicted grades:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE):", mae)
print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Final Grade (G3)")
plt.ylabel("Predicted Final Grade")
plt.title("Actual vs Predicted Final Grades")

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)

plt.savefig("visualizations/actual_vs_predicted.png")

plt.show()
```

```python
# 8. Random Forest Regression
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
rf_pred = rf_model.predict(X_test)

print("Random Forest predictions:")
print(rf_pred[:10])
```

```python
# 9. Model Evaluation
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest MAE:", rf_mae)
print("Random Forest RMSE:", rf_rmse)
print("Random Forest R² Score:", rf_r2)
```

```python
# 10. Model Comparison
model_comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [mae, rf_mae],
    "RMSE": [rmse, rf_rmse],
    "R2 Score": [r2, rf_r2]
})

print(model_comparison)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    model_comparison["Model"],
    model_comparison["R2 Score"]
)

plt.xlabel("Model")
plt.ylabel("R² Score")
plt.title("Model Performance Comparison")

plt.savefig("visualizations/model_comparison_r2.png")

plt.show()
```

# Student Performance Prediction Using Machine Learning

### Objective

The objective of this project is to predict students' final academic performance using student-related and academic features from the UCI Student Performance dataset.

### Machine Learning Models

- Linear Regression
- Random Forest Regression

### Target Variable

G3 — Final Grade

```python
# 14. Conclusion

This project developed a machine learning approach for predicting students' final grades using the UCI Student Performance dataset.

Two regression models were evaluated: Linear Regression and Random Forest Regression.

Linear Regression achieved an MAE of 1.647, an RMSE of 2.378, and an R² score of 0.724.

Random Forest achieved an MAE of 1.165, an RMSE of 1.949, and an R² score of 0.815 on the test set.

In this experiment, the Random Forest model produced lower prediction errors and a higher R² score than Linear Regression.

The results demonstrate that machine learning can be used to model relationships between student characteristics, previous academic performance, and final grades.

However, the results should be interpreted within the limitations of the dataset and experimental setup. In particular, G1 and G2 are included as input features, meaning that this model uses previous academic grades when predicting G3. Therefore, the model should not be described as an early-warning system that predicts performance before previous grades are available.
```
